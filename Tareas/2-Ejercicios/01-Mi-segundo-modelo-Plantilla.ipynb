## <font color='264CC7'> Introducción </font>

Este cuaderno está diseñado para practicar el entrenamiento y evaluación de un modelo de regresión lineal utilizando datos reales. Los objetivos principales son:

- Entrenar un modelo de regresión lineal.
- Evaluar el ajuste del modelo.


Del conjunto de datos disponible [aquí](https://raw.githubusercontent.com/mGalarnyk/Tutorial_Data/master/King_County/kingCountyHouseData.csv), se desea predecir la variable `price` en función de las variables `bedrooms`, `bathrooms`, `sqft_living`, `sqft_lot`, `floors`.

```python
# Paquetes necesarios
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

---
## <font color='264CC7'> Lectura de Datos </font>


Realiza la lectura de los datos y presenta las primeras filas del conjunto de datos.

```python
# Definir la URL de los datos
url = "https://raw.githubusercontent.com/mGalarnyk/Tutorial_Data/master/King_County/kingCountyHouseData.csv"

# Leer los datos con pandas
df = pd.read_csv(url)

# Filtrar el dataframe con las variables especificadas en la introducción
columnas = ['price', 'bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot', 'floors']
df_seleccion = df[columnas]

# Presentar las primeras filas del conjunto de datos
df_seleccion.head()
```

---
## <font color='264CC7'> Características y etiquetas </font>

Separa las características y las etiquetas y mostrar la dimensión de cada elemento.

```python
# Separar características (X) y etiquetas (y)
X = df_seleccion.drop(columns=['price'])
y = df_seleccion['price']

# Mostrar las dimensiones
print("Dimensión de las características (X):", X.shape)
print("Dimensión de las etiquetas (y):", y.shape)
```

---
## <font color='264CC7'> Conjunto de entrenamiento y prueba </font>

Divide los datos en dos conjuntos: uno para entrenamiento y otro para prueba y visualiza la dimensión de cada conjunto. Usar `random_state` igual al último número de tu cédula.

```python
# REEMPLAZA EL 0 POR EL ÚLTIMO NÚMERO DE TU CÉDULA
ultimo_numero_cedula = 0

# Dividir los datos (80% entrenamiento y 20% prueba es el estándar)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=ultimo_numero_cedula)

# Visualizar la dimensión de cada conjunto
print("Dimensión de X_train:", X_train.shape)
print("Dimensión de X_test:", X_test.shape)
print("Dimensión de y_train:", y_train.shape)
print("Dimensión de y_test:", y_test.shape)
```

Revisa la distribución de las etiquetas en los conjuntos de entrenamiento y prueba.

```python
# Al ser una variable numérica (precio), usamos describe() para ver su distribución estadística
print("--- Distribución de precios en Entrenamiento (y_train) ---")
print(y_train.describe())

print("\n--- Distribución de precios en Prueba (y_test) ---")
print(y_test.describe())
```

---
## <font color='264CC7'> Preprocesamiento de datos </font>

Decide si hace falta escalar las características para este ejemplo de regresión lineal. Si aplicas una transformación vista en clase, aprende sus parámetros solo con X_train y aplica la misma transformación a X_test, sin modificar la variable objetivo. No elijas transformaciones por su error de prueba.

```python
from sklearn.preprocessing import StandardScaler

# Inicializar el escalador
scaler = StandardScaler()

# Aprender los parámetros (fit) SOLO con X_train y aplicar la transformación
X_train_scaled = scaler.fit_transform(X_train)

# Aplicar la misma transformación a X_test sin recalcular los parámetros
X_test_scaled = scaler.transform(X_test)

print("Datos escalados correctamente.")
```

---
## <font color='264CC7'> Entrenamiento </font>

Define un modelo de regresión lineal y entrénalo con los datos de entrenamiento.

```python
# Definir el modelo de regresión lineal
modelo = LinearRegression()

# Entrenar el modelo con los datos de entrenamiento ya escalados
modelo.fit(X_train_scaled, y_train)

print("Modelo entrenado exitosamente.")
```

---
## <font color='264CC7'> Evaluación </font>

Realiza las predicciones con el conjunto de prueba y muestra los valores en una tabla.

```python
# Realizar las predicciones con el conjunto de prueba escalado
y_pred = modelo.predict(X_test_scaled)

# Crear un DataFrame para mostrar los valores reales frente a los predichos en una tabla
tabla_predicciones = pd.DataFrame({
    'Valor Real (y_test)': y_test,
    'Predicción (y_pred)': y_pred
})

# Mostrar las primeras filas de la tabla
tabla_predicciones.head()
```

Calcula el error cuadrático medio.

```python
# Calcular el error cuadrático medio (MSE)
mse = mean_squared_error(y_test, y_pred)

print(f"Error Cuadrático Medio (MSE): {mse}")
```
